```python

import random

n = int(input("Enter number of collection points (excluding depot): "))
n += 1

print("Enter the distance matrix row by row:")
distance = []
for i in range(n):
    row = list(map(int, input().split()))
    distance.append(row)

num_ants = 10
iterations = 100
alpha = 1.0
beta = 2.0
evaporation = 0.5
Q = 100

locations = ["Depot"] + [f"Point {i}" for i in range(1, n)]

pheromone = [[1.0] * n for _ in range(n)]
best_route = None
best_distance = float("inf")

for _ in range(iterations):
    all_routes = []

    for _ in range(num_ants):
        route = [0]
        unvisited = list(range(1, n))
        current = 0

        while unvisited:
            weights = []
            for city in unvisited:
                tau = pheromone[current][city] ** alpha
                eta = (1 / distance[current][city]) ** beta
                weights.append(tau * eta)

            next_city = random.choices(
                unvisited, weights=weights, k=1
            )[0]

            route.append(next_city)
            unvisited.remove(next_city)
            current = next_city

        route.append(0)

        total = sum(
            distance[route[i]][route[i + 1]]
            for i in range(len(route) - 1)
        )
        all_routes.append((route, total))

        if total < best_distance:
            best_distance = total
            best_route = route.copy()

    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - evaporation)

    for route, total in all_routes:
        deposit = Q / total
        for i in range(len(route) - 1):
            a, b = route[i], route[i + 1]
            pheromone[a][b] += deposit
            pheromone[b][a] += deposit

print("\nBest Garbage Collection Route:")
print(" -> ".join(locations[i] for i in best_route))
print("Total Distance:", best_distance)
```
